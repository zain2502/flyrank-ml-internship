## 2. The Question

### Research Question

**Which pages should be reviewed first for a possible content refresh or improvement based on their available search-performance and content signals?**

### Decision

The decision I want to support is **which pages should receive attention first** when a team has limited time for content review.

### Action

After getting a ranked list, a human reviewer could inspect the highest-priority pages and decide whether to refresh the content, improve it, monitor it, protect it, or leave it unchanged.

### Unit of Analysis

My main unit of analysis will be a **content page**. Each row represents a page/content item with measurements related to search performance and content characteristics.

### Output

The intended output is a **ranked review queue**. Each page could have a score or priority level and, later in the project, simple reason codes explaining why it was ranked highly.

### Cost of a Wrong Recommendation

A false positive could waste a team's time by sending someone to review a page that does not really need attention. A false negative could be more costly if an important page with a genuine opportunity or decline is missed.

Because of this, I do not want the model to automatically decide that a page must be changed. I want it to support a human reviewer by helping them decide where to look first.

### Why Data or ML Can Help

There can be a large number of pages to review, so manually checking every page can take a lot of time. Data can help narrow the list by looking at signals such as impressions, clicks, CTR, average position, content age, word count, and recent performance.

A simple rule can be a useful starting point because it is easy to understand. However, a machine learning model may be able to identify combinations of signals that are difficult to capture using a few fixed rules.

I already saw this idea in the starter notebook. The hand-written baseline achieved a Precision@50 of about 0.240, while the random forest reached about 0.740 on the starter evaluation. This suggests that a learned model can potentially improve the ranking of pages compared with the simple baseline on this particular dataset.

However, I do not see this as proof that machine learning will always be better. The purpose of the next few weeks is to test the idea carefully, define the target properly, check for leakage, and see whether the resulting recommendations are actually useful.

So this project is not just about training a model. The real goal is to help a person make a better decision about **which pages should be reviewed first**.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))
print("\nMedian impressions (90d):", round(df["impressions_90d"].median(), 2))
print("Percentage of pages with a downward trend:", round(df["trend_direction"].eq("down").mean() * 100, 2), "%")
print("Average CTR:", round(df["ctr"].mean(), 4))
```

```python
print("Average position:", round(df["avg_position"].mean(), 2))
print("Pages with at least 500 impressions:", int((df["impressions_90d"] >= 500).sum()))
```

### What I Observed

The starter dataset gives me enough page-level information to investigate a ranking and prioritization problem. The number of observations shows that there are many content items to consider, while the impressions, CTR, position, and trend measurements show that pages have different levels of search visibility and performance.

The starter pipeline also gave me a useful first comparison. The simple hand-written rule achieved about **0.240 Precision@50**, while the random forest achieved about **0.740 Precision@50** on the starter evaluation. This is a meaningful reason for me to investigate this lane further.

I am treating these results as evidence from the starter dataset rather than a guarantee about future data. The next step is to understand which signals are useful, define the decision carefully, and test whether the ranking remains useful under proper validation.

```python
print("Pages by trend direction:")
print(df["trend_direction"].value_counts())

print("\nMedian CTR by trend direction:")
print(df.groupby("trend_direction")["ctr"].median().round(4))
```

### Additional Observation

Looking at the data by trend direction gives me another reason to investigate this problem. The pages do not all behave the same way, and their search-performance measurements vary across the dataset.

This does not tell me that a particular signal causes a page to decline. It only gives me a starting point for investigating whether the available signals can help rank pages for human review.

## 4. Careful Words: What I Can and Can't Claim

### What I Can Claim

I can say that the starter dataset contains observable differences between pages in areas such as search impressions, CTR, position, content characteristics, and trend direction.

I can also compare different approaches using a clearly defined evaluation metric and see whether one approach performs better than another on the available evaluation data.

The starter results give me evidence that the learned model performed better than the hand-written baseline on Precision@50 in this particular experiment.

### What I Cannot Claim

I cannot claim that a particular feature is definitely a Google ranking factor just because it is related to another measurement in this dataset.

I also cannot claim that changing a page will definitely make its traffic or rankings improve. The data is observational, so relationships in the data do not automatically prove cause and effect.

I also need to be careful about future information. If I eventually build a model that makes a future-looking recommendation, I must make sure that features do not contain information from the future period I am trying to predict.

My goal is therefore to make evidence-based recommendations while being clear about what the data can and cannot tell me.

## Why This Is Not Just "Train a Model"

The main purpose of this project is not simply to train a model and report its accuracy. A useful ML project should start with a real decision.

In my case, the decision is **which pages should a content or SEO team review first?**

The model is only one part of the process. I first need to define the question, understand the data, decide what the target means, create a simple baseline, check for leakage, validate the results, and finally turn the results into recommendations that a person can actually use.

A model that performs well but does not help someone make a useful decision would not solve the actual problem.
